# Transliterate Text — Azure AI Translator

**Transliteration** converts text from one *script* to another, preserving the original pronunciation. The meaning of the text does not change — only the writing system used to represent it.

Example: Chinese Simplified `这是个测试。` → Latin script `zhè shì gè cèshì.`

```python
import os
from azure.ai.translation.text import TextTranslationClient
from azure.core.credentials import AzureKeyCredential
from azure.core.exceptions import HttpResponseError
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())  # loads .env from repo root

api_key = os.environ["FOUNDRY_AI_SERVICES_KEY"]
region = os.environ["FOUNDRY_AI_SERVICES_REGION"]

credential = AzureKeyCredential(api_key)
client = TextTranslationClient(credential=credential, region=region)
print("Translator client ready.")
```

```python
# Transliterate Chinese Simplified (Hans script) → Latin script
try:
    chinese_text = ["这是个测试。"]
    response = client.transliterate(
        body=chinese_text,
        language="zh-Hans",
        from_script="Hans",
        to_script="Latn",
    )
    result = response[0] if response else None
    if result:
        print(f"Original text  : {chinese_text[0]}")
        print(f"Transliterated : {result.text} (script: {result.script})")

except HttpResponseError as exc:
    if exc.error:
        print(f"Error {exc.error.code}: {exc.error.message}")
    raise
```
