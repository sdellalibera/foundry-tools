# Get Supported Languages — Azure AI Translator

This notebook queries the Azure Translator service for the full list of languages available for translation and transliteration.

No authentication key is required for this endpoint — it is publicly accessible.

```python
import os
from azure.ai.translation.text import TextTranslationClient
from azure.core.credentials import AzureKeyCredential
from azure.core.exceptions import HttpResponseError
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())  # loads .env from repo root

api_key = os.environ["AZURE_AI_KEY"]
region = os.environ["AZURE_AI_REGION"]

credential = AzureKeyCredential(api_key)
client = TextTranslationClient(credential=credential, region=region)
print("Translator client ready.")
```

```python
# Retrieve the full list of supported languages
try:
    response = client.get_supported_languages()

    translation_count = len(response.translation) if response.translation else 0
    transliteration_count = len(response.transliteration) if response.transliteration else 0

    print(f"Languages supported for translation   : {translation_count}")
    print(f"Languages supported for transliteration: {transliteration_count}")

except HttpResponseError as exc:
    if exc.error:
        print(f"Error {exc.error.code}: {exc.error.message}")
    raise
```

```python
# Print the first 20 translation languages
if response.translation:
    print("\nFirst 20 translation languages:")
    for code, lang in list(response.translation.items())[:20]:
        print(f"  {code:8s} — {lang.name} ({lang.native_name})")
```

```python
# Print the transliteration languages and their scripts
if response.transliteration:
    print("\nTransliteration languages:")
    for code, lang in response.transliteration.items():
        script_names = ", ".join(s.name for s in lang.scripts) if lang.scripts else "(none)"
        print(f"  {code:8s} — {lang.name} | scripts: {script_names}")
```
