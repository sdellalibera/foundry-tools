# OCR — Extract Text from Images

This notebook uses **Azure AI Vision Image Analysis 4.0** to extract printed and handwritten text from an image (OCR).

The `READ` visual feature returns bounding polygons and confidence scores for every line and word detected in the image.

```python
import os
from azure.ai.vision.imageanalysis import ImageAnalysisClient
from azure.ai.vision.imageanalysis.models import VisualFeatures
from azure.core.credentials import AzureKeyCredential
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())  # loads .env from repo root

endpoint = os.environ["AZURE_VISION_ENDPOINT"]
api_key = os.environ["AZURE_VISION_KEY"]

client = ImageAnalysisClient(endpoint=endpoint, credential=AzureKeyCredential(api_key))
print("Vision client ready.")
```

```python
# Use a publicly accessible image containing text
IMAGE_URL = "https://learn.microsoft.com/azure/ai-services/computer-vision/media/quickstarts/presentation.png"

print(f"Running OCR on: {IMAGE_URL}")

result = client.analyze_from_url(
    image_url=IMAGE_URL,
    visual_features=[VisualFeatures.READ],
)

print("OCR complete.")
```

```python
# Display all detected lines and words
if result.read and result.read.blocks:
    print("\nExtracted text:\n")
    for block in result.read.blocks:
        for line in block.lines:
            print(f"  Line: {line.text}")
            for word in line.words:
                print(f"    Word: '{word.text}' (confidence: {word.confidence:.4f})")
else:
    print("No text detected in the image.")
```

```python
# Print only the full text (all lines joined)
if result.read and result.read.blocks:
    all_lines = []
    for block in result.read.blocks:
        for line in block.lines:
            all_lines.append(line.text)
    full_text = "\n".join(all_lines)
    print("\n=== Full Extracted Text ===")
    print(full_text)
```
